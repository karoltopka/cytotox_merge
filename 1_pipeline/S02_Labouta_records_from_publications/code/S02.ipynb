# S02 – Labouta records from publications already covered by Gul et al. removed (DOI overlap)

```python
import sys

import numpy as np
import pandas as pd

sys.path.insert(0, ".")
from S02_comments import Audit, apply_comments   # the code of the comments of the step, apart from the data handling

INPUT = "../../S01_Raw_sources_loaded_and_mapped/S01.xlsx"
OUTPUT = "../S02.xlsx"
COMMENTS = "../S02_comments.xlsx"
AUDIT = "../S02_audit.xlsx"
STEP = "S02"
GUL, LABOUTA, NANOPHAROS = "Gul. Et al..", "Labouta Et. Al. ", "Nanopharos"

df = pd.read_excel(INPUT, keep_default_na=False, na_values=[""]).astype(object)
print(len(df), "records before the step")
audit = Audit(STEP, df, before=INPUT)
put, drop = audit.put, audit.drop                  # a rule sets a value / removes records through the audit, which records the change
changes, removed = audit.changes, audit.removed    # the ledger of the step
```

### 1. Records that only Labouta et al. have, checked against the publication figures: kept

```python
# sphere data that only Labouta has (Gul has only the quantum dots of this publication); kept by the C
KEEP = []
# 10.3762/bjnano.6.28 Fig. 1a (spheres): the Labouta value matches the bar of the spheres; Table 1 'spheres'
KEEP += [
    'L1832', 'L1833', 'L1836', 'L1837', 'L1840', 'L1841',
]
# 10.3762/bjnano.6.28 Fig. 1b (spheres): the Labouta value matches the bar of the spheres; Table 1 'spheres'
KEEP += [
    'L1164', 'L1165', 'L1166', 'L1167',
]
# 10.3762/bjnano.6.28 Fig. 3b (10 nm): the Labouta value matches the bar of the spheres; Table 1 'spheres'
KEEP += [
    'L1168', 'L1169', 'L1170', 'L1171', 'L1172',
]
# 10.3762/bjnano.6.28 Fig. 3b (24 nm): the Labouta value matches the bar of the spheres; Table 1 'spheres'
KEEP += [
    'L1173', 'L1174', 'L1175', 'L1176', 'L1177',
]
# 10.3762/bjnano.6.28 Fig. 1c (spheres): the Labouta value matches the bar of the spheres; Table 1 'spheres'
KEEP += [
    'L577', 'L578', 'L579', 'L580',
]
```

### 2. Labouta records of publications already covered by Gul et al.: removed

```python
gul_dois = set(df.loc[df._src == GUL, "_doi"])
mask = (df._src == LABOUTA) & df._doi.isin(gul_dois) & ~df.raw_id.isin(KEEP)
drop(mask, f"{STEP}_overlap", "Labouta record from a publication already covered by Gul et al. (DOI overlap) - Gul record kept")
print(int(mask.sum()), "records removed")
```

### 3. Comments of the step

```python
# the comments table of this step is empty: the step takes no decision on the strength of a publication
apply_comments(audit, COMMENTS, [])
```

### The table after the step and its audit

```python
# the table after the step; the audit module compares it with the table before the step and writes the audit workbook
df.to_excel(OUTPUT, index=False)
audit.write(AUDIT)
```
