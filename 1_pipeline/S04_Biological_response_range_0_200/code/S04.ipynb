# S04 – Biological response range 0-200 % viability

```python
import sys

import numpy as np
import pandas as pd

sys.path.insert(0, ".")
from S04_comments import Audit, apply_comments   # the code of the comments of the step, apart from the data handling

INPUT = "../../S03_Records_without_publication_determinable/S03.xlsx"
OUTPUT = "../S04.xlsx"
COMMENTS = "../S04_comments.xlsx"
AUDIT = "../S04_audit.xlsx"
STEP = "S04"

df = pd.read_excel(INPUT, keep_default_na=False, na_values=[""]).astype(object)
print(len(df), "records before the step")
audit = Audit(STEP, df, before=INPUT)
put, drop = audit.put, audit.drop                  # a rule sets a value / removes records through the audit, which records the change
changes, removed = audit.changes, audit.removed    # the ledger of the step
```

### 1. Viability values missing, below 0 % or above 200 %: removed

```python
v = pd.to_numeric(df["Cell_Viability (%)"], errors="coerce")
drop(v.isna(), f"{STEP}_viability_missing", "viability value missing / not numeric")
v = pd.to_numeric(df["Cell_Viability (%)"], errors="coerce")
drop(v < 0, f"{STEP}_viability_below_0", "viability < 0 % - experimental error or extraction artefact")
v = pd.to_numeric(df["Cell_Viability (%)"], errors="coerce")
drop(v > 200, f"{STEP}_viability_above_200", "viability > 200 % - outside the interpretable range of moderate hormetic stimulation")
print(len(removed), "records removed")
```

### 2. Comments of the step

```python
# the comments table of this step is empty: the step takes no decision on the strength of a publication
apply_comments(audit, COMMENTS, [])
```

### The table after the step and its audit

```python
# the table after the step; the audit module compares it with the table before the step and writes the audit workbook
df.to_excel(OUTPUT, index=False)
audit.write(AUDIT)
```
