# S05 – Chemical scope: inorganic metals and metal oxides (no organic, carbon, HAP, QD)

```python
import sys

import numpy as np
import pandas as pd

sys.path.insert(0, ".")
from S05_comments import Audit, apply_comments   # the code of the comments of the step, apart from the data handling

INPUT = "../../S04_Biological_response_range_0_200/S04.xlsx"
OUTPUT = "../S05.xlsx"
COMMENTS = "../S05_comments.xlsx"
AUDIT = "../S05_audit.xlsx"
STEP = "S05"

df = pd.read_excel(INPUT, keep_default_na=False, na_values=[""]).astype(object)
print(len(df), "records before the step")
audit = Audit(STEP, df, before=INPUT)
put, drop = audit.put, audit.drop                  # a rule sets a value / removes records through the audit, which records the change
changes, removed = audit.changes, audit.removed    # the ledger of the step
```

### 1. Organic and polymeric materials: removed

```python
drop(df["Type"].astype(str).str.strip().eq("O"), f"{STEP}_organic", "organic / polymeric nanomaterial - outside inorganic scope")
print(len(removed), "records removed")
```

### 2. Material classes outside the metal and metal-oxide scope: removed

```python
EXCLUDED = {   # reason -> material names of the source files
    'carbon allotrope - outside metal / metal-oxide scope': [
        'C60', 'C70', 'Carbon', 'Carbon NP', 'Carbon Nanotubes', 'Diamond',
        'Graphite', 'MWCNT', 'SWCNT',
    ],
    'hydroxyapatite excluded (calcium/lysosomal mechanism, small class)': [
        'HAP', 'Hydroxyapatite',
    ],
    'quantum dots excluded (semiconductor nanocrystals, separate OECD class)': [
        'QD', 'QDs',
    ],
}
n = len(removed)
for reason, materials in EXCLUDED.items():
    drop(df["_mat_raw"].isin(materials), f"{STEP}_material_class", reason)
print(len(removed) - n, "records removed")
```

### 3. Comments of the step

```python
# the comments table of this step is empty: the step takes no decision on the strength of a publication
apply_comments(audit, COMMENTS, [])
```

### The table after the step and its audit

```python
# the table after the step; the audit module compares it with the table before the step and writes the audit workbook
df.to_excel(OUTPUT, index=False)
audit.write(AUDIT)
```
