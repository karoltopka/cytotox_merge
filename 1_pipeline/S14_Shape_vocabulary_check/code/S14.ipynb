# S14 – Shape vocabulary check (one label: Sphere, assigned at S07) and the publication's description of the shape (Shape_Comments)

```python
import sys

import numpy as np
import pandas as pd

sys.path.insert(0, ".")
from S14_comments import Audit, apply_comments   # the code of the comments of the step, apart from the data handling

INPUT = "../../S13_Particle_size_harmonization/S13.xlsx"
OUTPUT = "../S14.xlsx"
COMMENTS = "../S14_comments.xlsx"
AUDIT = "../S14_audit.xlsx"
STEP = "S14"

df = pd.read_excel(INPUT, keep_default_na=False, na_values=[""]).astype(object)
print(len(df), "records before the step")
audit = Audit(STEP, df, before=INPUT)
put, drop = audit.put, audit.drop                  # a rule sets a value / removes records through the audit, which records the change
changes, removed = audit.changes, audit.removed    # the ledger of the step
```

### 1. Shape vocabulary: one label for spherical particles

```python
SHAPE_LABELS = {'Spherical': 'Sphere'}
for i in df.index:
    s = str(df.at[i, "Shape"]).strip()
    if s in SHAPE_LABELS:
        put(i, "Shape", SHAPE_LABELS[s], f"{STEP}_shape_label",
            "shape vocabulary: one label for spherical particles (Labouta 'Spherical' = Gul/NanoPharos 'Sphere')", outside=False)
```

### 2. Values from the publication review: the comments of this step

```python
# every value from the publication review: the rows of these rules in the comments table (record, variable, new value or removal, comment)
apply_comments(audit, COMMENTS, ['S14_curated_L1_doi', 'S14_curated_L4_diameter', 'S14_curated_L7_row'])
```

### The table after the step and its audit

```python
# the table after the step; the audit module compares it with the table before the step and writes the audit workbook
df.to_excel(OUTPUT, index=False)
audit.write(AUDIT)
```
