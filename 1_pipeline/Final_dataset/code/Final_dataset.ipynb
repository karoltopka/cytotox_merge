# Final dataset

```python
import pandas as pd

INPUT = "../../S24_Serum_status_of_the_exposure_medium/S24.xlsx"
OUTPUT = "../dataset_final.xlsx"
AUDIT = "../Final_dataset_audit.xlsx"
COLUMNS = ['raw_id', 'Material', 'Shape', 'Shape_Comments', 'Coat/Functional Group', 'Synthesis_Method', 'Surface_Charge', 'Dry_Diameter (nm)', 'Dry_Diameter_Source', 'Dry_Diameter_Method', 'DLS_Diameter_Water (nm)', 'DLS_Diameter_Medium (nm)', 'DLS_Diameter_Other (nm)', 'Size_Measurement_Comments', 'Zeta_in_Water (mV)', 'Zeta_in_Medium (mV)', 'Zeta_in_Other (mV)', 'Cell_Type', 'Cancer', 'Cell_organ', 'Cell_type_family', 'Germ_layer', 'No_of_Cells (cells/well)', 'Time (hr)', 'Concentration (ug/ml)', 'Concentration (g/l)2', 'Concentration (uM)', 'Concentration_Comments', 'Serum_Status', 'Test', 'Test_family', 'Test_Indicator', 'Cell_Viability (%)', 'DOI', 'Animal_mapped_combined', 'Dataset']

df = pd.read_excel(INPUT, keep_default_na=False, na_values=[""]).astype(object)
dropped = [c for c in df.columns if c not in COLUMNS]
df[COLUMNS].to_excel(OUTPUT, index=False)
pd.DataFrame({"column not published": dropped}).to_excel(AUDIT, index=False)
print(len(df), "records |", len(COLUMNS), "published columns |", len(dropped), "columns not published")
```
